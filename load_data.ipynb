```python
import torch
import torchvision
import numpy as np
from torch.utils.data import ConcatDataset
import os
import pillow_heif
from PIL import Image
from tqdm import tqdm
from concurrent.futures import ThreadPoolExecutor, as_completed

mnist_train = torchvision.datasets.MNIST(root="./data", train=True, download=True)
mnist_test = torchvision.datasets.MNIST(root="./data", train=False, download=True)

combined_dataset = ConcatDataset([mnist_train, mnist_test])
train_len = len(mnist_train)

os.makedirs("data/mnist", exist_ok=True)

def save_heif(i):
    img, cls = combined_dataset[i]
    if cls == 10:
        return
    array = np.uint8(np.array(img))
    heif_file = pillow_heif.from_pillow(Image.fromarray(array))
    subset = "train" if i < train_len else "test"
    class_dir = f"data/mnist/{cls}"
    os.makedirs(class_dir, exist_ok=True)
    filename = f"{subset}_{i}.heif"
    heif_file.save(f"{class_dir}/{filename}")

with ThreadPoolExecutor(max_workers=8) as executor:
    futures = [executor.submit(save_heif, i) for i in range(len(combined_dataset))]
    for _ in tqdm(as_completed(futures), total=len(futures)):
        pass
```

### MNIST Rows

```python
import torch
import torchvision
import numpy as np
from torch.utils.data import ConcatDataset
import os
import pillow_heif
from PIL import Image
from tqdm import tqdm
from concurrent.futures import ThreadPoolExecutor, as_completed

mnist_train = torchvision.datasets.MNIST(root="./data", train=True, download=True)
mnist_test = torchvision.datasets.MNIST(root="./data", train=False, download=True)

combined_dataset = ConcatDataset([mnist_train, mnist_test])
train_len = len(mnist_train)

os.makedirs("data/mnist_rows", exist_ok=True)

# create the 4 on, 4 off repeating pattern mask for 28 rows
pattern_mask = np.ones((28, 28), dtype=np.uint8)
for start_row in range(0, 28, 8):
    off_start = start_row + 4
    off_end = min(start_row + 8, 28)
    pattern_mask[off_start:off_end, :] = 0


def save_heif(i):
    img, cls = combined_dataset[i]

    if cls == 10:
        return
    
    array = np.uint8(np.array(img))
    array = np.uint8(array * pattern_mask)

    heif_file = pillow_heif.from_pillow(Image.fromarray(array))
    subset = "train" if i < train_len else "test"
    class_dir = f"data/mnist_rows/{cls}"
    os.makedirs(class_dir, exist_ok=True)
    filename = f"{subset}_{i}.heif"
    heif_file.save(f"{class_dir}/{filename}")

with ThreadPoolExecutor(max_workers=8) as executor:
    futures = [executor.submit(save_heif, i) for i in range(len(combined_dataset))]
    for _ in tqdm(as_completed(futures), total=len(futures)):
        pass
```
